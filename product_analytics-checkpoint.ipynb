```python
import pandas as pd
import matplotlib.pyplot as plt
from sqlalchemy import create_engine
from urllib.parse import quote_plus

pw = quote_plus("YOUR_PASSWORD")  # your postgres password
engine = create_engine(f"postgresql+psycopg2://postgres:{pw}@localhost:5432/ecommerce_analytics")

df = pd.read_sql("""
    SELECT user_id, product_id, event_type, event_time, price, main_category
    FROM events_clean
""", engine)

df["event_time"] = pd.to_datetime(df["event_time"], utc=True)
print(df.shape)
```

```python
first_view = df[df.event_type == "view"].groupby("user_id")["event_time"].min()
first_buy = df[df.event_type == "purchase"].groupby("user_id")["event_time"].min()

t = pd.concat([first_view.rename("first_view"), first_buy.rename("first_buy")], axis=1).dropna()
t["hours"] = (t["first_buy"] - t["first_view"]).dt.total_seconds() / 3600
t = t[t["hours"] >= 0]  # drop users whose purchase is logged before their first view

print("Users:", len(t), "| median hours:", round(t.hours.median(), 2))

labels = ["within 1 hour", "1-24 hours", "1-7 days", "over 7 days"]
bins = [-0.001, 1, 24, 168, 1e9]
dist = pd.cut(t.hours, bins=bins, labels=labels).value_counts(normalize=True).reindex(labels) * 100
print(dist.round(1))

ax = dist.plot(kind="bar", figsize=(7, 4), color="#2a6fb0")
ax.set_title("How long users take to buy after their first view")
ax.set_ylabel("% of buyers")
ax.set_xlabel("")
plt.xticks(rotation=0)
for i, v in enumerate(dist):
    ax.text(i, v + 1, f"{v:.1f}%", ha="center")
plt.savefig("time_to_purchase.png", dpi=150, bbox_inches="tight")
plt.show()
```

```python
cart = df[df.event_type == "cart"][["user_id", "product_id", "main_category", "price"]] \
         .drop_duplicates(["user_id", "product_id"])
buy = df[df.event_type == "purchase"][["user_id", "product_id"]].drop_duplicates()
buy["bought"] = 1

c = cart.merge(buy, on=["user_id", "product_id"], how="left")
c["abandoned"] = 1 - c["bought"].fillna(0)

print(f"Cart items: {len(c):,} | abandoned: {int(c.abandoned.sum()):,} ({c.abandoned.mean()*100:.1f}%)")

by_cat = c.groupby("main_category").agg(carts=("abandoned", "size"), abandon_pct=("abandoned", "mean"))
by_cat["abandon_pct"] = (by_cat.abandon_pct * 100).round(1)
by_cat = by_cat[by_cat.carts >= 500].sort_values("abandon_pct", ascending=False)
print(by_cat)

c["price_band"] = pd.cut(c.price, [0, 50, 200, 500, 1e9], labels=["under 50", "50-200", "200-500", "over 500"])
by_price = c.groupby("price_band", observed=True).agg(carts=("abandoned", "size"), abandon_pct=("abandoned", "mean"))
by_price["abandon_pct"] = (by_price.abandon_pct * 100).round(1)
print(by_price)

ax = by_price["abandon_pct"].plot(kind="bar", figsize=(7, 4), color="#c0504d")
ax.set_title("Cart abandonment by item price")
ax.set_ylabel("% of cart items never purchased")
ax.set_xlabel("")
plt.xticks(rotation=0)
plt.savefig("cart_abandonment_by_price.png", dpi=150, bbox_inches="tight")
plt.show()
```
